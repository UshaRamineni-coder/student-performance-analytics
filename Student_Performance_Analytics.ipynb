```python

# Select the subjects
subjects = ["Python", "Mathematics", "Database", "Communication"]

# Calculate total marks
df["Total_Marks"] = np.sum(df[subjects].to_numpy(), axis=1)

# Calculate average marks
df["Average_Marks"] = np.mean(df[subjects].to_numpy(), axis=1)

# Display the results
display(df[["Name", *subjects, "Total_Marks", "Average_Marks"]])
```

- subjects stores the names of the four subject columns.
- to_numpy() converts the subject marks into a NumPy array.
- np.sum(..., axis=1) calculates the total marks for each student.
- np.mean(..., axis=1) calculates each student's average marks.
- display() shows the results in a table.
Since there are four subjects, the maximum total is 400 if each subject is marked out of 100.

```python

print("Total students:", len(df))
print("Class average:", round(df["Average_Marks"].mean(), 2))
print("Highest average:", round(df["Average_Marks"].max(), 2))
print("Lowest average:", round(df["Average_Marks"].min(), 2))
```

These calculations help summarize the class performance.

```python

# Function to calculate student grades
def calculate_grade(average):
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    elif average >= 60:
        return "D"
    elif average >= 50:
        return "E"
    else:
        return "F"


# Apply the function to every student
df["Grade"] = df["Average_Marks"].apply(calculate_grade)

# Display student grades
display(df[["Name", "Average_Marks", "Grade"]])
```

- def creates a function named calculate_grade.
- if and elif check the student's average marks.
- return gives back the appropriate grade.
- .apply(calculate_grade) applies the function to every student's average.
- df["Grade"] stores the grades in a new column.
These grade boundaries are the rules we're choosing for this project.

```python

def calculate_result(row):
    if row[subjects].min() >= 35 and row["Average_Marks"] >= 40:
        return "PASS"
    else:
        return "FAIL"


df["Result"] = df.apply(calculate_result, axis=1)

display(df[["Name", "Average_Marks", "Grade", "Result"]])
```

Pass rule: A student must score at least 35 in every subject and have an average of at least 40.

```python

# Class performance statistics

total_students = len(df)

class_average = df["Average_Marks"].mean()

highest_average = df["Average_Marks"].max()

lowest_average = df["Average_Marks"].min()

pass_count = (df["Result"] == "PASS").sum()

fail_count = (df["Result"] == "FAIL").sum()

pass_percentage = (pass_count / total_students) * 100

average_attendance = df["Attendance_Percent"].mean()

print("===== CLASS PERFORMANCE REPORT =====")
print("Total Students:", total_students)
print("Class Average:", round(class_average, 2))
print("Highest Average:", round(highest_average, 2))
print("Lowest Average:", round(lowest_average, 2))
print("Students Passed:", pass_count)
print("Students Failed:", fail_count)
print("Pass Percentage:", round(pass_percentage, 2), "%")
print("Average Attendance:", round(average_attendance, 2), "%")
```

len() counts students.



.mean() calculates the average.



.max() and .min() find the highest and lowest values.



.sum() counts how many students passed or failed.



Pass percentage shows the proportion of students who passed.

```python

print("===== SUBJECT-WISE PERFORMANCE =====")

for subject in subjects:
    print("\nSubject:", subject)
    print("Average Marks:", round(df[subject].mean(), 2))
    print("Highest Marks:", df[subject].max())
    print("Lowest Marks:", df[subject].min())

subject_averages = df[subjects].mean()

print("\nBest Performing Subject:", subject_averages.idxmax())
print("Lowest Performing Subject:", subject_averages.idxmin())
```

This uses a for loop to analyze all four subjects without repeating the same code manually

```python

# Find the top 3 students based on average marks

top_students = df.nlargest(3, "Average_Marks")

print("===== TOP 3 PERFORMING STUDENTS =====")

display(
    top_students[
        ["Student_ID", "Name", "Department",
         "Average_Marks", "Grade"]
    ]
)
```

nlargest(3, "Average_Marks") selects the three students with the highest average marks.

```python

# Find students who scored above 80 in Python

high_python_scores = df[df["Python"] > 80]

print("===== STUDENTS SCORING ABOVE 80 IN PYTHON =====")

display(
    high_python_scores[
        ["Name", "Python", "Average_Marks", "Grade"]
    ]
)
```

This uses a Pandas condition to filter students whose Python marks are greater than 80.

# Student Performance Analytics System

## 1. Project Objective

The objective of this project is to analyze student academic performance using Python, NumPy, and Pandas. The system calculates total marks, average marks, grades, and pass/fail results. It also analyzes subject-wise performance and identifies the top three students.

## 2. Technologies Used

- **Python:** Used for programming logic, functions, loops, and conditional statements.
- **NumPy:** Used for calculating total and average marks.

- **Google Colab:** Used to write and execute Python code online.

The dataset contains student information, subject marks, and attendance percentages. It is used to analyze academic performance and generate student results.

## 4. Project Features

1. Load and display the student dataset.
2. Calculate total marks for each student.
3. Calculate average marks for each student.
4. Assign grades based on average marks.
5. Classify students as PASS or FAIL.
6. Calculate overall class performance statistics.
7. Analyze subject-wise average, highest, and lowest marks.
8. Identify the top three performing students.
9. Filter students who scored above 80 in Python.
10. Save the analyzed results as a CSV file.

## 5. Expected Outcome

The system generates a structured analysis of student performance, helps identify high-performing students and weaker subjects, and exports the results to a CSV file for further use.
